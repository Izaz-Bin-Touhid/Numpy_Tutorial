```python
#NdArray

arr1 = np.array([1,2,3])
print("1D Array",arr1)

arr2 = np.array([[1,2,3],[4,5,6]])
print("2D array", arr2)

arr3 = np.array([
    [
        [1,2,3],
        [3,5,6]
    ],
    [
        [3,4,6],
        [7,8,2]
    ],
])
print("3D array", arr3)
print(arr3.shape)
```

```python
#ndarray Creating from scratch

arr = np.zeros((2,3), dtype=np.int8)
print (arr)
print(type(arr))
print(arr.shape)
print(arr.ndim)
print(arr.size)

arr = np.zeros_like(arr)
print (arr)
print(type(arr))
print(arr.shape)
print(arr.ndim)
print(arr.size)

arr = np.ones((2,3), dtype= np.int8)
print (arr)
print(type(arr))
print(arr.shape)
print(arr.ndim)
print(arr.size)

arr = np.ones_like(arr)
print (arr)
print(type(arr))
print(arr.shape)
print(arr.ndim)
print(arr.size)

arr = np.empty ((4,3), dtype=np.int16)
print (arr)
print(type(arr))
print(arr.shape)
print(arr.ndim)
print(arr.size)

arr = np.empty_like (arr)
print (arr)
print(type(arr))
print(arr.shape)
print(arr.ndim)
print(arr.size)

arr = np.full ((4,3),np.inf)
print (arr)
print(type(arr))
print(arr.shape)
print(arr.ndim)
print(arr.size)

arr = np.full_like(arr,np.inf, dtype=np.float64)
print (arr)
print(type(arr))
print(arr.shape)
print(arr.ndim)
print(arr.size)
```

```python
# Array creation with random values

arr = np.random.rand(2,3) #np.random.rand(shape)
print(arr)
print(arr.shape)
print(type(arr))
print(arr.ndim)
print(arr.size)

# np.random.rand(start, end, (shape))
arr = np.random.randint(1,10, (2,3))
print(arr)
print(arr.shape)
print(type(arr))
print(arr.ndim)
print(arr.size)

```

```python
# Array creation with range function

array = np.arange(1,10,1)
matrix = array.reshape(3,3)

print(array)
print(matrix)
print(array.shape)
print(array.dtype)
print(type(array))
print(array.size)

array = np.linspace( 0 ,4,5)
print(array)
print(matrix)
print(array.shape)
print(array.dtype)
print(type(array))
print(array.size)

array = np.logspace(0,4,6, base=3)
print(array)
print(matrix)
print(array.shape)
print(array.dtype)
print(type(array))
print(array.size)
```

```python
diagonal_matrix = np.diag([1,2,3,4,])
print(diagonal_matrix)
print(diagonal_matrix.shape)

identity_matrix = np.eye(4)
print(identity_matrix)
print(identity_matrix.shape)

matrix = np.eye(3,4,1)
print(matrix)
```

```python

# Indexing and Slicing
arr1 = np.array([10, 20, 30, 40, 50])
arr2 = np.array([[1, 2, 3], 
                 [4, 5, 6]])

print(arr1)
print(arr2)

arr1[1]= 200
print(arr1)

arr2[1][2]= 50
print(arr2)

# 1D array Slicing : 1d = arr[start:end:step]
arr1_mod = arr1[1:4]. copy()
arr1[2]=4

arr1_mod[2]=30
print(arr1_mod)

```
