```python
#ndarray Creating from scratch

arr = np.zeros((2,3), dtype=np.int8)
print (arr)
print(type(arr))
print(arr.shape)
print(arr.ndim)
print(arr.size)

arr = np.zeros_like(arr)
print (arr)
print(type(arr))
print(arr.shape)
print(arr.ndim)
print(arr.size)

arr = np.ones((2,3), dtype= np.int8)
print (arr)
print(type(arr))
print(arr.shape)
print(arr.ndim)
print(arr.size)

arr = np.ones_like(arr)
print (arr)
print(type(arr))
print(arr.shape)
print(arr.ndim)
print(arr.size)

arr = np.empty ((4,3), dtype=np.int16)
print (arr)
print(type(arr))
print(arr.shape)
print(arr.ndim)
print(arr.size)

arr = np.empty_like (arr)
print (arr)
print(type(arr))
print(arr.shape)
print(arr.ndim)
print(arr.size)

arr = np.full ((4,3),np.inf)
print (arr)
print(type(arr))
print(arr.shape)
print(arr.ndim)
print(arr.size)

arr = np.full_like(arr,np.inf, dtype=np.float64)
print (arr)
print(type(arr))
print(arr.shape)
print(arr.ndim)
print(arr.size)
```

```python
# Array creation with range function

array = np.arange(1,10,1)
matrix = array.reshape(3,3)

print(array)
print(matrix)
print(array.shape)
print(array.dtype)
print(type(array))
print(array.size)

array = np.linspace( 0 ,4,5)
print(array)
print(matrix)
print(array.shape)
print(array.dtype)
print(type(array))
print(array.size)

array = np.logspace(0,4,6, base=3)
print(array)
print(matrix)
print(array.shape)
print(array.dtype)
print(type(array))
print(array.size)
```

```python

# Indexing and Slicing
arr1 = np.array([10, 20, 30, 40, 50])
arr2 = np.array([[1, 2, 3], 
                 [4, 5, 6]])

print(arr1)
print(arr2)

arr1[1]= 200
print(arr1)

arr2[1][2]= 50
print(arr2)

# 1D array Slicing : 1d = arr[start:end:step]
arr1_mod = arr1[1:4]. copy()
arr1[2]=4

arr1_mod[2]=30
print(arr1_mod)

```

```python
#Concatenate
#Row wise concatenate er shomoy column same thakbe
#Column wise concatenate er shomoy row same thakbe

a = np.random.randint(1,10, size=(2,3))
print(a)
b = np.random.randint(10,100, size=(2,3))
print(b)

con_row = np.concatenate((a,b), axis=0) # for column concatenation axis will be 1
print(con_row)
```
